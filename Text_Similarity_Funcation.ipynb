```python
from recordlinkage.datasets import load_febrl1
dfa = load_febrl1()
dfa.head()
```

#Blocking

```python
from recordlinkage.index import Block
```

```python
indexer = Block('given_name')
pair = indexer.index(dfa)
print(len(pair))
```

#similarty Matcching and Scoring

```python
dfa.head()
```

```python
import recordlinkage
com = recordlinkage.Compare()
```

```python
com.string('given_name','given_name',method='jarowinkler',label='given_name')
com.string('surname','surname',method='jarowinkler',label='surname')
com.exact('date_of_birth','date_of_birth',label='date_of_birth')
com.exact('state','state',label='state')
com.string('address_1','address_1',method='jarowinkler',label='address_1')
fe = com.compute(pair,dfa)
fe.sample(5)
```

#predict the match using ECM Classifier

```python
fes = fe[['given_name','surname','date_of_birth','state','address_1']]
ecm = recordlinkage.ECMClassifier()
re = ecm.learn((fes).astype(int),return_type='series')
re
```
